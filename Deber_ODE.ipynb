Ejercicio 1

```python
import numpy as np
import matplotlib.pyplot as plt

# Método de Euler
def euler_method(f, t0, y0, h, n_steps):
    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)
    t[0] = t0
    y[0] = y0
    for i in range(n_steps):
        y[i + 1] = y[i] + h * f(t[i], y[i])
        t[i + 1] = t[i] + h
    return t, y

# Definir las funciones derivadas para el Ejercicio 1
def f1(t, y):
    return t * np.exp(3 * t) - 2 * y

def f2(t, y):
    return 1 + (t - y)**2

def f3(t, y):
    return 1 + y/t

def f4(t, y):
    return np.cos(2 * t) + np.sin(3 * t)

# Condiciones iniciales y parámetros para el Ejercicio 1
h_values = [0.5, 0.5, 0.25, 0.25]
t_intervals = [(0, 1), (2, 3), (1, 2), (0, 1)]
y0_values = [0, 1, 2, 1]
functions = [f1, f2, f3, f4]

# Resolver las ecuaciones del Ejercicio 1
solutions_1 = []
for h, (t0, t_end), y0, f in zip(h_values, t_intervals, y0_values, functions):
    n_steps = int((t_end - t0) / h)
    t, y = euler_method(f, t0, y0, h, n_steps)
    solutions_1.append((t, y))

# Graficar las soluciones del Ejercicio 1
plt.figure(figsize=(14, 8))
for i, (t, y) in enumerate(solutions_1, start=1):
    plt.subplot(2, 2, i)
    plt.plot(t, y, marker='o', label=f'Función {i}')
    plt.xlabel('t')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)
plt.suptitle('Método de Euler - Ejercicio 1')
plt.tight_layout()
plt.show()
```

```python
# Método de Taylor de orden 2
def taylor_order_2(f, df, t0, y0, h, n_steps):
    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)
    t[0] = t0
    y[0] = y0
    for i in range(n_steps):
        y[i + 1] = y[i] + h * f(t[i], y[i]) + (h**2 / 2) * df(t[i], y[i])
        t[i + 1] = t[i] + h
    return t, y

# Definir la derivada de f1(t, y) para Taylor de orden 2 en el Ejercicio 6
def df1(t, y):
    return 3*t*np.exp(3*t) - 2*f1(t, y)

# Aplicar el método de Taylor de orden 2 para el problema (a)
t, y = taylor_order_2(f1, df1, 0, 0, 0.5, 2)
print("Solución de Taylor de orden 2 para el problema (a) del ejercicio 6:")
print(y)
```

```python
# Método de Taylor de orden 4
def taylor_order_4(f, df, ddf, dddf, t0, y0, h, n_steps):
    t = np.zeros(n_steps + 1)
    y = np.zeros(n_steps + 1)
    t[0] = t0
    y[0] = y0
    for i in range(n_steps):
        y[i + 1] = (y[i] + h * f(t[i], y[i]) + (h**2 / 2) * df(t[i], y[i]) +
                    (h**3 / 6) * ddf(t[i], y[i]) + (h**4 / 24) * dddf(t[i], y[i]))
        t[i + 1] = t[i] + h
    return t, y

# Se puede definir df, ddf, y dddf según las derivadas de las funciones de los ejercicios.
# Aquí voy a resolver solo para el primer caso del Ejercicio 7, que sería una continuación del Ejercicio 6.

# Definir derivadas para el Ejercicio 7
def ddf1(t, y):
    return 9 * t * np.exp(3 * t) - 2 * df1(t, y)

def dddf1(t, y):
    return 27 * t * np.exp(3 * t) - 2 * ddf1(t, y)

# Aplicar el método de Taylor de orden 4 para el problema (a)
t, y = taylor_order_4(f1, df1, ddf1, dddf1, 0, 0, 0.5, 2)
print("Solución de Taylor de orden 4 para el problema (a) del ejercicio 7:")
print(y)
```
